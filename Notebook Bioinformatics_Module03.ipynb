# **<font color=blue>P4. </font>** Data types: use ```type()``` to find a variable's "type"
####**TASK P4.1:** Ask Gemini, "*Give me examples of each data type to see how they look as outputs.*"

### Integer Data Type
Integers are whole numbers, positive or negative, without decimals.

````python
int = 100
print(f"Value: {int}, Type: {type(int)}")
````

### Float Data Type
Floats are numbers with decimal points.

````python
float = 10.5
print(f"Value: {float}, Type: {type(float)}")
````

### String Data Type
Strings are sequences of characters, enclosed in single or double quotes.

````python
str = "Hello Colab"
print(f"Value: {str}, Type: {type(str)}")
````

### Boolean Data Type
Booleans represent truth values, either `True` or `False`.

````python
bool = True
print(f"Value: {bool}, Type: {type(bool)}")
````

### List Data Type
Lists are ordered, mutable collections of items, enclosed in square brackets `[]`.

````python
list = [1, 2, 3, "apple"]
print(f"Value: {list}, Type: {type(list)}")
````

### Tuple Data Type
Tuples are ordered, immutable collections of items, enclosed in parentheses `()`.

````python
tup = (1, 2, 3, "banana")
print(f"Value: {tup}, Type: {type(tup)}")
````

### Dictionary Data Type
Dictionaries are unordered collections of key-value pairs, enclosed in curly braces `{}`.

````python
dict = {"name": "Alice", "age": 30}
print(f"Value: {dict}, Type: {type(dict)}")
````

### Set Data Type
Sets are unordered collections of unique items, enclosed in curly braces `{}`.

````python
set = {1, 2, 3, 2}
print(f"Value: {set}, Type: {type(set)}")
````
